```python
while True:
    print("""
          1. Add element
          2. Remove element
          3. Display first 3 elements
          4.Display last 3 elements
          5.print all elements
          6. Exit  
          """)
    ch=int(input("Enter your choice: "))
    if ch==1:
        x=int(input("Enter a number: "))
        l1.append(x)    
    elif ch==2:
        x=int(input("Enter a number: "))
        if x in l1:
            l1.remove(x)
        else:
            print("Element not found in list")
    elif ch==3:
        print(l1[0:3])
    elif ch==4:
        print(l1[-3:])
    elif ch==5:
        if len(l1)==0:
            print("empty list  ")
        print(l1)
    elif ch==6:
        break   
```

```python
#recursion
def factorial(n):
    if n==0:
        return 1
    else:
        return n*factorial(n-1)
    
print(factorial(5))
```

```python
#python csv
import csv

l=[["Name", "Age", "City"],
   ["Devesh", 20, "Mumbai"]]

with open ("data.csv","w", newline='') as file:
    writer =csv.writer(file)

    for i in l:
        writer.writerow(i)
```
