```python
#concat df
import pandas as pd
data1=pd.DataFrame({"Name":['Alice','Bob','Charlie'],
                    "Age":[25,30,35]})
data2=pd.DataFrame({"Name":['David','Eve','Frank'],
                    "Age":[40,45,50]})

res=pd.concat([data1,data2],axis=0,ignore_index=True)
print(res)
```

```python
#Merge
#syntax DF.merge(DF2,on='column_name',how='inner')
right_df=pd.DataFrame({"ID":[1,2,3],
                       "Name":['Alice','Bob','Charlie']})
left_df=pd.DataFrame({"ID":[1,2,4],
                      "Age":[25,30,35]})
res=right_df.merge(left_df,on='ID',how='inner')
print(res)
```

```python
#pyhton pandas pivoting
import pandas as pd

df = pd.DataFrame({
    'Col1':range(12),
    'Col2':['A']*3 + ['B']*3 + ['C']*3 + ['D']*3,
    'Date': pd.to_datetime(['2026-01-01', '2026-01-02', '2026-01-03']*4)
})
print(df)
print("\n\npivoted table\n")
pivot_table = df.pivot_table(index='Date', columns='Col2', values='Col1')

print(pivot_table)
```

```python
data = {
    "Course":['CS','CIVIL','CS','CIVIL','CS'],
    "ID":[1,2,1,1,2],
    'Student':[100,20,150,120,90]
}
df = pd.DataFrame(data)
print(df)


print("\n\npivoted table\n")
pivoted=df.pivot_table(index="Course",columns="ID",values="Student",aggfunc='sum')
print(pivoted)

df['marks']= [95,15,8,4,45]
print(df)

print(df['Student'][3])
```
